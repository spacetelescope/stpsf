# Important note about pupil orientation

All the phase maps distributed in stpsf-data use the Roman entrance pupil orientation. This is for consistency with the preferred orientation of the optical data provide by NASA. The Roman PSF simulation, however, uses the Roman exit pupil orientation. This matches the Roman PSF as observed in the detector frame of reference. 

The figure below shows the different optical planes inside a Roman STPSF simulation. 

```python
import stpsf 
import matplotlib.pylab as plt 


plt.figure(figsize = [12,12])
wfi = stpsf.WFI()
wfi.filter = 'F062'
wfi.detector = 'WFI07'
fov_pixels = 256
psf, planes = wfi.calc_psf( fov_pixels = fov_pixels, return_intermediates=True, display = True)
```

Alternatively we can show only the pupil intensity, first and second panels from the figure above.  

```python
fig, (ax1, ax2) = plt.subplots(1,2,figsize=(14,7), sharey=True)
planes[0].display(ax = ax1, title = f'Roman Entrance Pupil \n {wfi.detector}')
planes[1].display(ax = ax2, title = f'Roman Exit Pupil \n {wfi.detector}')
```

> **⚠️ Important Note**
>
> STPSF automatically takes care of flipping the pupil orientation as part of the calculation. Users of STPSF should generally not have to take any action related to this. However, users who may wish to reuse the pupil masks and optical data as part of their own independent optical modeling must keep in mind these coordinate conventions.
